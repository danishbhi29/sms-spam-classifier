## 1️⃣ Environment Setup
Python environment aur required libraries import karna.

```python
import pandas as pd
import string

print('✅ Libraries loaded successfully!')
```

---
## 2️⃣ Dataset
SMS Spam Collection dataset load karna. Yeh tab-separated file hai jismein 2 columns hain: `label` (ham/spam) aur `message` (SMS text).

```python
df = pd.read_csv('../data/SMSSpamCollection', sep='\t', header=None, names=['label', 'message'])

print(f'Dataset Shape: {df.shape}')
print(f'Total Messages: {len(df)}')
print(f'Columns: {list(df.columns)}')
print()
df.head()
```

## 3️⃣ Data Exploration
Dataset ko achi tarah samajhna — class distribution, missing values, duplicates, aur text characteristics.

### 3.1 Basic Info & Statistics

```python
df.info()
df.describe()
```

### 3.2 Class Distribution (Ham vs Spam)

```python
print('--- Class Distribution ---')
print(df['label'].value_counts())
print()
print('--- Percentage ---')
print(df['label'].value_counts(normalize=True) * 100)
print()
# Agar model har message ko HAM predict kare, tab bhi uski accuracy ~87% hogi!
# Isliye hum baad mein sirf accuracy par depend nahi karenge.
print('⚠️ Dataset is IMBALANCED — ~87% ham, ~13% spam')
print('💡 Solution: class_weight="balanced" use karenge model training mein')
```

### 3.3 Missing Values & Duplicates

```python
print(f'Missing Values: {df.isnull().sum().sum()}')
print(f'Duplicate Rows: {df.duplicated().sum()}')

# Remove duplicates
df = df.drop_duplicates()
print(f'\n✅ After removing duplicates:')
print(f'Remaining Rows: {df.shape[0]}')
print(f'Duplicates Now: {df.duplicated().sum()}')
print()
print('--- Updated Class Distribution ---')
print(df['label'].value_counts(normalize=True) * 100)
```

## 4️⃣ Data Preprocessing
Text clean karna — lowercase, punctuation remove, extra spaces remove.

```python
# Step 1: Lowercase
df['message'] = df['message'].str.lower()

# Step 2: Remove punctuation
df['message'] = df['message'].str.translate(
    str.maketrans('', '', string.punctuation)
)

# Step 3: Remove extra spaces
df['message'] = df['message'].str.replace(r'\s+', ' ', regex=True).str.strip()

print('✅ Preprocessing complete!')
print()
df.head()
```

### 4.2 Label Encoding
Label ko text (ham/spam) se numbers (0/1) mein convert karna — ham = 0, spam = 1.

```python
# Encode labels: ham = 0, spam = 1
df['label'] = df['label'].map({'ham': 0, 'spam': 1})

print('✅ Label encoding complete!')
print()
print(df['label'].value_counts())
print()
df.head()
```
